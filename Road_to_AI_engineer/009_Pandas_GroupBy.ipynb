```python
genres = movies.groupby('Genre')

def rank_movie(group):
  group['genre_rank'] = group['IMDB_Rating'].rank(ascending=False)
  return group

genres.apply(rank_movie, include_groups=False)
```

```python
def normal(group):
  group['normalise'] = (group['IMDB_Rating']-group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group

genres.apply(normal)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("deliveries.csv")

# Filter only Virat Kohli's batting data
kohli_df = df[df['batsman'] == 'V Kohli']

# Group by opponent team (bowling team)
grouped = kohli_df.groupby('bowling_team')

# Calculate stats
grouped.agg(
    matches=('match_id', 'nunique'),
    innings=('match_id', 'count'),
    runs=('batsman_runs', 'sum'),
    balls=('ball', 'count'),
    fours=('batsman_runs', lambda x: (x == 4).sum()),
    sixes=('batsman_runs', lambda x: (x == 6).sum())
)

# # Calculate strike rate
# result['strike_rate'] = (result['runs'] / result['balls']) * 100

# # Sort by runs
# result = result.sort_values(by='runs', ascending=False)

# print(result)
```

```python
# Create a function that can return the highest score of any batsman

def highest_score(batsman):
  mask = ipl[ipl['batsman'] == batsman]
  return mask.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1)

highest_score('V Kohli')
```

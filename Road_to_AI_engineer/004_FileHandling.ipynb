```python
# Serialozing and deserializing custom objects

class Person:
  def __init__(self, fname, lname, age, gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

# format to printed in
# -> Nitish Singh age -> 33 gender -> male
```

```python
from textwrap import indent
person = Person('Chirag', 'Bansal', 22, 'Male')

import json
def show_object(person):
  if isinstance(person, Person):
    return '{} {} age -> {} gender -> {}'.format(person.fname, person.lname, person.age, person.gender)

with open('demo3.json', 'w') as f:
  json.dump(person, f, default=show_object)
```

```python
# As a dict

def show_object(person):
  if isinstance(person, Person):
    return {
        'name' : person.fname + " " + person.lname,
        'age' : person.age,
        'gender' : person.gender
    }

with open('demo4.json', 'w') as f:
  json.dump(person, f, indent=4, default=show_object)
```

```python
# Pickle
class Person:

  def __init__(self, name, age, gender):
    self.name = name
    self.age = age
    self.gender = gender

  def show_info(self):
    print('Hi, my name is ', self.name, 'I am ', self.age, " year old")

person = Person('Chirag Bansal', 22, 'Male')

import pickle
with open('peron.pkl', 'wb') as f:
  pickle.dump(person, f)
```

```python
with open('peron.pkl', 'rb') as f:
   p = pickle.load(f)
```

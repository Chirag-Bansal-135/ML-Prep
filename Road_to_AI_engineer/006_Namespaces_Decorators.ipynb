```python
# Global and Local scope
a = 2     # global

def temp():
  a = 3   # local
  print(a)

temp()
print(a)
```

```python
# local and global -> editing global
a = 2

def temp():
  a+= 1
  print(a)

temp()

print(a)
```

```python
a = 2

def temp():
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# Enclosing namespace

def outer():
  a = 3          # enclosing
  def inner():
    print('inner function')
    a = 2            # local
    print(a)
  inner()
  print('Outer function')
a = 1          # global
outer()
print('main program')
```

```python
# nonlocal keyword
def outer():
  a = 3          # enclosing
  def inner():
    print('inner function')
    nonlocal a
    a += 2            # local
    print(a)
  inner()
  print('Outer function')
a = 1          # global
outer()
print('main program')
```

```python
# Pyhton are 1st class function

def modify(func, num):
  return func(num)

def square(num):
  return num**2

modify(square, 2)
```

```python
 # Decorators

def my_decorator(func):
  def wrapper():
    print("******************")
    func()
    print('******************')
  return wrapper

def hello():
  print('Hello')

a = my_decorator(hello)
a()
```

```python
# how to use decorator
def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

@my_decorator
def hello():
  print('Hello')

hello()
```

```python
import time

def timer(func):
  def wrapper():
    start = time.time()
    func()
    print('time taken is', time.time()-start, 'secs')
  return wrapper

@timer
def hello():
  print('hello')
  time.sleep(2)

hello()
```

```python
import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken', func.__name__, time.time()-start, 'secs')
  return wrapper

@timer
def hello():
  print('Hello')

@timer
def square(num):
  time.sleep(1)
  print(num**2)

hello()
square(7)
```

```python
 # Check data type

def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(args[0]) == data_type:
        func(*args)
      else:
        raise TypeError("wrong DataType")
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

square(7)
```

```python
# Aggregation
class Customer:

  def __init__(self, name, gender, address):
    self.name = name
    self.gender = gender
    self.address = address

  def print_address(self):
    print(self.address.get_city(), self.address.pin, self.address.state)

  def edit_profile(self, new_name, new_city, new_pin, new_state):
    self.name = new_name
    self.address.edit_address(new_city, new_pin, new_state)

class Address:

  def __init__(self, city, pin, state):
    self.__city = city
    self.pin = pin
    self.state = state

  def get_city(self):
    return self.__city

  def edit_address(self, new_city, new_pin, new_state):
    self.__city = new_city
    self.pin = new_pin
    self.state = new_state

add1 = Address('Patiala', 147002, 'Punjab')
cust1 = Customer('Chirag', 'Male', add1)

print(cust1.address.pin)
cust1.print_address()

cust1.edit_profile('Naresh', 'Parwanoo', 173220, 'Himachal')

cust1.print_address()
```

```python
# child can't access private members of the class
class phone:
  def __init__(self, price, brand, camera):
    print("Inside phone constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def show(self):
    print(self.__price)

class smartphone(phone):
  def check(self):
    print(self.__price)

s = smartphone(20000, "samsung", 13)
s.show()

```

```python
# Method Overriding
class Phone:
  def __init__(self, price, brand, camera):
    print("Inside phone constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print('Buying a phone')

class Smartphone(phone):
  def buy(self):
    print("Buying a samrtphone")

s = Smartphone(10000, 'mi', '8mp')
s.buy()
```

```python
# super -> constructor
class Phone:
  def __init__(self, price, brand, camera):
    print('Inside the phone constructor')
    self.price = price
    self.brand = brand
    self.camera = camera

class Smartphone(Phone):
  def __init__(self, price, camera, brand, os, ram):
    print('Inside smartphone constructor')
    super().__init__(price, brand, camera)
    self.os = os
    self.ram = ram
    print('Inside samrtphone constructor')

s = Smartphone(20000, 8, 'apple', 'macos', 4)
print(s.os)
print(s.brand)
```

```python
#Abstraction
from abc import ABC, abstractmethod

class BankApp(ABC):
  def database(self):
    print('Connected to DataBase')

  @abstractmethod
  def secruity(self):
    pass
```

```python
class MobileApp(BankApp):
  def mobile_login(self):
    print('login into mobile')

  def secruity(self):
    print('mobile secruity')
```

```python
mob = MobileApp()
mob.secruity()
```

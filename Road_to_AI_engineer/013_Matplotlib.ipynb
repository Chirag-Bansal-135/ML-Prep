```python
batters = pd.read_csv('/content/batter.csv')
# batters = batters.head(25)
```

```python
# Annotation graph
plt.figure(figsize=(18,10))

plt.scatter(batters['avg'], batters['strike_rate'], s=batters['runs'])
plt.xlabel('average')
plt.ylabel('strike rate')

plt.axhline(130, color='red')
plt.axvline(35, color='red')

for i in range(batters.shape[0]):
  plt.text(batters['avg'].values[i], batters['strike_rate'].values[i], batters['batter'].values[i])
```

```python
 # Subplots
fig, ax = plt.subplots(figsize=(15,6))

ax.scatter(batters['avg'], batters['strike_rate'], color='red', marker='+')
ax.set_title('blah blah blah')
ax.set_xlabel('avg')
ax.set_ylabel('striker Rate')

fig.show()
```

```python
# for multiple subplots
fig, ax = plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(10,6))

ax[0].scatter(batters['avg'], batters['strike_rate'], color='red')
ax[1].scatter(batters['avg'], batters['runs'])

ax[0].set_title('avg vd strike rate')
ax[1].set_title('avg vs runs')

ax[1].set_xlabel('avg')
# ax[1].set_xlabel('avg')
```

```python
fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,6))

ax[0,0].scatter(batters['avg'], batters['strike_rate'], color='red')
ax[1,0].scatter(batters['avg'], batters['runs'])
ax[0,1].hist(batters['strike_rate'])
ax[1,1].hist(batters['runs'])
```

```python
# surface plot

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)

z = xx**2 + yy**2

fig = plt.figure()

ax = fig.add_subplot(projection='3d')
ax.plot_surface(xx,yy,z)
```

```python
x = np.linspace(1,10,10)
y = np.linspace(11, 20,10)
xx, yy = np.meshgrid(x,y)

yy * xx
```

```python
z = np.sin(xx) + np.sin(yy)

fig = plt.figure()

ax = fig.add_subplot(projection='3d')
p = ax.plot_surface(xx,yy,z, cmap='viridis')
fig.colorbar(p)
```

```python
# contour
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx, yy = np.meshgrid(x,y)

z = xx**2 + yy**2

fig = plt.figure(figsize=(12,8))

ax = fig.add_subplot()
p = ax.contour(xx,yy,z)
fig.colorbar(p)
```

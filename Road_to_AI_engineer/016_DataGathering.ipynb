```python
#header parameter
pd.read_csv('/content/test (1).csv', header=1).head()
# header - make row of that number into columns of Dataframe
```

```python
# skiprows / nrows Parameter
pd.read_csv('aug_train.csv', skiprows=[1,2,4])
# skiprows=[] - don't add the index provided in the list
# nrows - specifies how many columns we need to read
```

```python
# Encoding parameter
pd.read_csv('zomato.csv', encoding='latin-1').head()
# encoding - if files are written in different format, we can specifies those format to read the files
```

```python
# on_bad_lines='warn' or 'skip'
# if lets say we have dataframe with different size of data in different
# rows, to avoid the error, we can use on_bad_lines parameter
pd.read_csv('BX-Books.csv', encoding='latin-1', on_bad_lines='warn')
```

```python
# dtypes parameter
pd.read_csv('aug_train.csv', dtype={'target':'int32'}).info()
```

```python
# convertor
def rename(name):
  if name == 'Royal Challengers Bangalore':
    return 'RCB'
  else:
    return name

pd.read_csv('/content/IPL Matches 2008-2020.csv', converters={'team1':rename}).head()
```

```python
# na_values parameter
pd.read_csv('aug_train.csv', na_values=['Male'])
# na_values - when unexpected value come in the columns which is neither
# make sense with column and not a NaN value. So its hard to work with
# those rows. We convert those rows in Nan useing na_values
```

## Pandas Export

```python
# to_csv
df = pd.read_csv('deliveries.csv')
df.head()
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df
```

```python
temp_df.to_csv('batsman_runs.csv', index=False)
```

```python
temp_df2 = df.pivot_table(index='batsman', columns='bowling_team', values='batsman_runs', aggfunc='sum')
temp_df2.to_csv('batsman_vs_team.csv')
```

```python
#to_excel
df
```

```python
temp_df.to_excel('batsman_runs.xlsx', index=False)
```

```python
df.query('batsman_runs == 6 and ball in [1,2,3,4,5,6]').pivot_table(index='over', columns='ball', values='batsman_runs', aggfunc='count').to_html('sixes_heatmap.html')
```

```python
# to JSON
df.groupby(['batting_team', 'batsman'])['batsman_runs'].sum().unstack().to_json('ipl.json')
```

## DF using Web Scraping

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
headers = {'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
url = 'https://www.ambitionbox.com/list-of-companies'

webpage = requests.get(url, headers=headers).text
```

```python
soup = BeautifulSoup(webpage, 'lxml')
```

```python
# print(soup.prettify())
```

```python
soup.find_all('h1')[0].text
```

```python
for i in soup.find_all('h2'):
    print(i.text.strip())
```

```python
soup.find_all('div', class_='rating_text rating_text--md ')
```

```python
soup.find("div", class_=["rating_text", "rating_text--md"]).text.strip()
```

```python
company = soup.find_all('div', class_='companyCardWrapper__primaryInformation')
len(company)
```

```python
name = []
rating = []
ctype = []
for i in company:
    name.append(i.find('h2', class_='companyCardWrapper__companyName').text.strip())
    rating.append(i.find("div", class_=["rating_text", "rating_text--md"]).text.strip())
    ctype.append(i.find('span', class_="companyCardWrapper__interLinking").text.strip())


df = pd.DataFrame({'name':name,
                  'rating':rating,
                  'company type':ctype})
```

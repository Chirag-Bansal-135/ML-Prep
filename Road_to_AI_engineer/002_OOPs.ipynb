### Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line



```python


class Point:

  def __init__(self, x, y):
    self.x_cod = x
    self.y_cod = y

  def __str__(self):
    return '<{},{}>'.format(self.x_cod, self.y_cod)

  def eculidian_distance(self, other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return self.eculidian_distance(Point(0,0))


class Line:

  def __init__(self, A, B, C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.A, self.B, self.C)

  def point_on_line (line, point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return 'lies on the line'
    else:
      return 'do not lie on the line'

  def shortest_distance(line, point):
    num = abs(line.A*point.x_cod + line.B*point.y_cod +line.C)
    den = (line.A + line.B)**0.5
    return num/den

```

```python
p1 = Point(1,1)
print(p1)
p2 = Point(2,3)
#p1.eculidian_distance(p2)
p1.distance_from_origin()
```

```python
l1 = Line(1,2,3)
l2 = Line(1,1,-2)
print(l1)
print(l2.point_on_line(p1))
print(l2.shortest_distance(p1))
```

```python
# Collection of objects
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person('Chirag', 22)
p2 = Person("Ram", 21)
p3 = Person("Shyam", 23)

L = [p1, p2, p3]
print(L)
S = (p1, p2, p3)
print(S)
for i in S:
    print(i.name)
p1.name = 'mohan'
print(L)
print(S)
for i in S:
    print(i.name)
```

```python
# Static Variable

class Count:
  counter = 0

  def __init__(self):
    self.num = Count.counter
    Count.counter += 1

num1 = Count()
num2 = Count()
num3 = Count()
print(num1.num)
print(num2.num)
print(num3.num)
print(Count.counter)
```

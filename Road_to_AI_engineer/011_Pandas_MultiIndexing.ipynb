```python
deaths = pd.read_csv('/content/time_series_covid19_deaths_global.csv')
deaths.head()
```

```python
confirmed = pd.read_csv('/content/time_series_covid19_confirmed_global.csv')
confirmed.head()
```

```python
dea = deaths.melt(id_vars=['Province/State', 'Country/Region', 'Lat', 'Long'], var_name='dates', value_name='deaths')
```

```python
cases = confirmed.melt(id_vars=['Province/State', 'Country/Region', 'Lat', 'Long'], var_name='dates', value_name='cases')
```

```python
data = cases.merge(dea, on=['Province/State', 'Country/Region', 'Lat', 'Long', 'dates'])
```

```python
data[['Country/Region','deaths', 'cases']]
```

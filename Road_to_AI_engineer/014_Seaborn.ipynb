```python
tips = sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axes level
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
# hue - distinguish those point in our graph by changing color
# style - further tell the information about those point by changing the marker shape
# size - change the size of marker to show their weightage
```

```python
# relplot -> figure level -> square shape
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', hue='sex', style='time', size='size')
```

Plots under distribution plot

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate Histogram
sns.histplot(tips, x='total_bill', bins=20)
#bins - distribut data in bins, more bins, more thoroughly data distribute
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
# hist plot using categorical data
sns.displot(data=tips, x='day', kind='hist')
```

```python
sns.displot(data=tips, x='tip', kind='hist', hue='sex')
```

```python
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
# element='step' - it removes the line in between those hist bar plot
```

```python
sns.displot(data=tips, x='tip', kind='hist', col='sex', element='step')
```

```python
# kdeplot
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
# fill
sns.displot(kind='kde', data=tips, x='total_bill', hue='sex', fill=True, height=10, aspect=2)
# height - Increse the graph height
# aspect - any number which we provide, it will increase width that times to the height
```

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips, x='total_bill')
sns.rugplot(data=tips, x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
```

### Figure level function -> `catplot`

```python
# strip plot
sns.stripplot(data=tips, x='day', y='total_bill', jitter=False)
# jitter=false - its make our data in a straight lint
```

```python
# figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', hue='sex')
```

```python
#swarmplot
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
```

```python
#boxplot
sns.boxplot(data=tips, x='sex', y='total_bill')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='box')
```

```python
sns.boxplot(data=tips, y='total_bill')
```

```python
# violinplot = (boxplot + KDEplot)
sns.violinplot(data=tips, x='day', y='total_bill')
```

```python
sns.catplot(data=tips, x='day', y='total_bill', kind='violin', hue='sex', split=True)
```

When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
# BAr plot
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.std, errorbar=None)
# errorbar= None - it will remove vertical line on bars
```

```python
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.max)
```

```python
# lineplot
sns.pointplot(data=tips, x='sex', y='total_bill', hue='smoker')
```

Countplot

A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
# countplot
sns.countplot(data=tips, x='sex', hue='day')
```

```python
# faceting using catplot
sns.catplot(data=tips, x='sex', y='total_bill', kind='box', col='smoker', row='time')
```

### Regression Plots

- regplot (axes level)
- lmplot (figure level)

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
# regplot -> axes level
sns.regplot(data=tips, x='total_bill', y='tip')
```

```python
#lmplot -> figure level
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

```python
# residplot
sns.residplot(data=tips, x='total_bill', y='tip')
# it tell us how much point deviated from original mark
```

```python
# A second way to plot facet plots -> FacetGrid
# sns.catplot(data=tips, x='sex', y='total_bill', kind='violin', col='day', row='time')

g = sns.FacetGrid(data=tips, col='day', row='time', hue='smoker')
g.map(sns.violinplot, 'sex', 'total_bill')
g.add_legend()
```

### Plotting Pairwise relationship (PairGrid Vs Pairplot)

```python
iris = iris.set_index('species_id')
sns.pairplot(iris, hue='species')
```

```python
g = sns.PairGrid(data=iris, hue='species')
g.map(sns.scatterplot)
```

```python
g = sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.violinplot)
g.map_offdiag(sns.histplot)
g.add_legend()
```

```python
g = sns.PairGrid(data=iris, hue='species', vars=['sepal_width','petal_width'])
g.map_diag(sns.violinplot)
g.map_upper(sns.histplot)
g.map_lower(sns.kdeplot)
g.add_legend()
```

### Jointplot Vs JointGrid

```python
sns.jointplot(data=tips, x='total_bill', y='tip', kind='hist', hue='sex')
```

```python
g = sns.JointGrid(data=tips, x='total_bill', y='tip', hue='sex')
g.plot(sns.scatterplot, sns.histplot)
```

```python
# Creating a function where if we 2 team name, it tell how many match they won

def win_comp(ipl, team1, team2):
  case1 = (ipl['Team1'] == team1) & (ipl['Team2'] == team2)
  case2 = (ipl['Team1'] == team2) & (ipl['Team2'] == team1)

  win_team_1 = ipl['WinningTeam'] == team1
  win_team_2 = ipl['WinningTeam'] == team2

  t1 = ipl[case1 | case2 & win_team_1].shape[0]
  t2 = ipl[case1 | case2 & win_team_2].shape[0]


  print('{} won {} match against {}'.format(team1, t1, team2))
  print('{} won {} match against {}'.format(team2, t2, team1))
```

```python
win_comp(ipl, 'Chennai Super Kings', 'Rajasthan Royals')
```
